```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters.character import CharacterTextSplitter
```

```python
loader=Docx2txtLoader("Introduction_to_Data_and_Data_Science_2.docx")
pages=loader.load()
```

```python
for i in range(len(pages)):
    pages[i].page_content = ' '.join(pages[i].page_content.split())
```

```python
len( pages[0].page_content )
```

```python
char_spilter=CharacterTextSplitter(separator=' ',chunk_size=500,chunk_overlap=10)
```

```python
pages_char_split=char_spilter.split_documents(pages)
```

```python
len(pages_char_split)
```

```python
for doc in pages_char_split:
    # Strip any trailing whitespace and add a period if it doesn't already have one
    text = doc.page_content.rstrip()
    if not text.endswith('.'):
        doc.page_content = text + '.'
```
